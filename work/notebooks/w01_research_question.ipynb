## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** Identifying groups of webpages with similar performance so that appropriate actions (protect, refresh, improve, merge, prune, or monitor) can be taken for each group. <br>
**Action:** Grouping webpages together which needs similar actions. <br>
**Cost of a Wrong Call:** If a page is assigned to the wrong archetype, it may receive the wrong recommendation. For example, if a page that needs a refresh is classified as a Champion and recommended for protection, outdated or incorrect information may remain on the website, reducing user trust and potentially harming search performance over time. Similarly, if a high-performing page is mistakenly recommended for pruning or rewriting, it could lose traffic and visibility.
